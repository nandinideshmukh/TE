```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = sns.load_dataset("titanic")
```

```python
data
```

```python
data = data.drop(columns=['sibsp','embark_town','embarked','parch','adult_male','class','alone','deck','who'],axis=1)
```

```python
data.columns
```

```python
from sklearn.preprocessing import LabelEncoder

lb = LabelEncoder()

columns_to_encode = [ 'sex', 'alive']
for col in columns_to_encode:
    data[col] = lb.fit_transform(data[col])
```

```python
data
```

```python
plt.figure(figsize=(8, 6)) 
sns.heatmap(data.corr(),annot=True)
plt.show()
```

```python
sns.histplot(data=data, x='fare',hue='pclass', kde=0, bins=30, color='green')
plt.title('Distribution of Passenger Fares')
plt.xlabel('Fare')
plt.ylabel('Frequency')
plt.show()
```

```python

plt.figure(figsize=(10,6))
sns.boxplot(x='pclass',y='fare',data = data)
plt.title('Fare distrubution with each passenger')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.show()
```

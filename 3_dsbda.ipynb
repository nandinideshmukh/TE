```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = pd.read_csv("/home/TE/Downloads/31116/3_dsbda/nba.csv")
data
```

```python
data.isna().sum()
```

```python
data.shape
```

```python
data = data.dropna(subset=['Name','Team','Number','Position','Age','Height','Weight','Salary'])
```

```python


data["College"] = data["College"].fillna(data["College"].mode())
data["Salary"] = data["Salary"].fillna(data["Salary"].mean())
data.dropna(inplace=True)

```

```python
data
```

```python
from sklearn.preprocessing import LabelBinarizer

lb = LabelBinarizer()
data['College'] = lb.fit_transform(data['College'])
data['Team'] = lb.fit_transform(data['Team'])
data['Position'] = lb.fit_transform(data['Position'])
data = data.drop(columns=['Name','Weight'])
```

```python
grouped_data = data.groupby(data['Height'])
grouped_data.head()
```

```python
grouped_data.describe()
```

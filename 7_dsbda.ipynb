```python
import nltk
try:
    nltk.download('punkt_tab')
except Exception:
    nltk.download('punkt')
```

```python
with open('7.doc', 'r') as file:
    doc = file.read()
```

```python
word_tokens = nltk.word_tokenize(doc)
```

```python
print(word_tokens)
```

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english')) 
word_tokens = [token for token in word_tokens if token not in stop_words]
print(word_tokens)
```

```python
nltk.download('averaged_perceptron_tagger_eng')
```

```python
from nltk import pos_tag
tagged = pos_tag(word_tokens)
print(tagged)
```

```python
nltk.download('wordnet')
```

```python
lemmatizer = nltk.stem.WordNetLemmatizer()
lemmatized_tokens = [lemmatizer.lemmatize(token) for token in word_tokens]
print(lemmatized_tokens)
```

```python
stemmer = nltk.stem.PorterStemmer()
stemmed_tokens = [stemmer.stem(token) for token in word_tokens]
print(stemmed_tokens)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
result = tfidf.fit_transform(lemmatized_tokens)
```

```python
print(result)
```

```python
print('\nidf values:')
for ele1, ele2 in zip(tfidf.get_feature_names_out(), tfidf.idf_):
    print(ele1, ':', ele2)
```

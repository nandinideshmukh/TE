```python
import pandas as pd

data = pd.read_csv("/home/TE/Downloads/31116/3_dsbda/Iris.csv")
data
```

```python
print(f"Columns in the database: \n{data.columns}")
print()
print(f"Datatypes of the column:\n{data.dtypes}")
```

```python
from sklearn.preprocessing import LabelBinarizer
lb = LabelBinarizer()
data['Species'] = lb.fit_transform(data['Species'])
```

```python
data
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

for column in data.columns:
    plt.title(f'Histogram for {column}')
    sns.histplot(data=data, x=column, bins=18, color='green') 
    plt.show()
```

```python
for column in data.columns:
    if column!='Species':
        plt.figure(figsize=(8, 5)) 
        plt.title(f'Boxplot for {column}')
        sns.boxplot(data=data, x=column, color='blue')
        plt.show()
```

```python
data.dtypes
```

```python
numeric_data = data.select_dtypes(include=['float64', 'int64'])

plt.figure(figsize=(8,6))
sns.heatmap(numeric_data.corr(), annot=True, cmap='coolwarm')
plt.show()
```
